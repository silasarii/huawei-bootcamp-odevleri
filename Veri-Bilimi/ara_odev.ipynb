##### **Soru 13 -- birim_fiyat sütunu için IQR yöntemini kullanarak Q1, Q3, IQR, alt sınır ve üst sınır değerlerini hesaplayın. Sonra aykırı değer içeren kayıt sayısını yazdırın.**

```python
Q1 = veri["birim_fiyat"].quantile(0.25)
Q3 = veri["birim_fiyat"].quantile(0.75)

IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR
```

```python
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Alt sınır:", alt_sinir)
print("Üst sınır:", ust_sinir)
```

```python
aykiri_deger_sayisi = (
    (veri["birim_fiyat"] < alt_sinir) |
    (veri["birim_fiyat"] > ust_sinir)
).sum()

aykiri_deger_sayisi
```

**Yorum:** birim_fiyat sütunu için IQR yöntemi uygulandığında Q1 değeri 198,855, Q3 değeri 817,335 ve IQR değeri 618,48 olarak hesaplanmıştır. Buna göre alt sınır -728,865, üst sınır ise 1745,055 olarak bulunmuştur. Bu sınırlar dışında kalan 92 kayıt aykırı değer olarak belirlenmiştir. Alt sınırın negatif olması nedeniyle aykırı değerlerin üst sınırın üzerinde yer aldığı görülmektedir.
